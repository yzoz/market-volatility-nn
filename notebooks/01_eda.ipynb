# Exploratory Data Analysis

This notebook inspects the synthetic OHLCV data and the engineered volatility
features. It runs without network access so it can be executed in CI or by anyone
who clones the repository.

```python
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import pandas as pd

from volnn.config import FeatureConfig
from volnn.data.features import build_feature_frame
from volnn.data.synthetic import generate_klines
```

```python
klines = generate_klines(n_steps=5000, seed=42)
cfg = FeatureConfig(ticker="DEMO", interval="5m", past=64, future=8, vol_window=4)
frame = build_feature_frame(pd.DataFrame(klines), cfg)
close = klines[:, 4]
hv = frame[:, 0]
frame.shape, close.shape, hv.shape
```

```python
fig, axes = plt.subplots(3, 1, figsize=(11, 9))
axes[0].plot(close)
axes[0].set_title("Synthetic close price")
axes[1].plot(hv)
axes[1].set_title("Realised volatility (hv)")
features = pd.DataFrame(frame, columns=["hv", "ac", "vol", "count"])
corr = features.corr()
im = axes[2].imshow(corr, cmap="coolwarm")
axes[2].set_xticks(range(4), features.columns)
axes[2].set_yticks(range(4), features.columns)
axes[2].set_title("Feature correlation")
fig.colorbar(im, ax=axes[2])
plt.tight_layout()
plt.show()
```
